# Import

```python
import pandas as pd
```

The project CSV files are stored in the data/raw

```python
ad_revenue = pd.read_csv('../data/raw/ad_revenue.csv')
iap_products = pd.read_csv('../data/raw/iap_products.csv')
matches = pd.read_csv('../data/raw/matches.csv')
purchases = pd.read_csv('../data/raw/purchases.csv')
sessions = pd.read_csv('../data/raw/sessions.csv')
ua_spend = pd.read_csv('../data/raw/ua_spend.csv')
users = pd.read_csv('../data/raw/users.csv')
```

Shapes

```python
files = ['ad_revenue', 'iap_products','matches', 'purchases', 'sessions', 
        'ua_spend', 'users']

for f in files: 
    df = pd.read_csv(f'../data/raw/{f}.csv')
    print(f'{f}: {df.shape}')
```

file: ad_revenue

```python
print(f'shape:{ad_revenue.shape}')
print(ad_revenue.dtypes)
print(ad_revenue.isnull().sum())
display(ad_revenue.head())

# Todo: Standardize categories
print(ad_revenue['tipo_ad'].value_counts())
print(ad_revenue['red_ad'].value_counts())
# revenue describe
print(ad_revenue['ingreso'].describe())
```

file: iap_products

```python
print(f'shape:{iap_products.shape}')
print(iap_products.dtypes)
print(iap_products.isnull().sum())
display(iap_products.head())

# Todo: Standardize categories
print(iap_products['tipo'].value_counts())
```

file: matches

```python
print(f'shape: {matches.shape}')
print(matches.dtypes)
print(matches.isnull().sum())
display(matches.head())

# Todo: Standardize categories
print(matches['resultado'].value_counts())

# Describe
print(matches['movimientos'].describe())
print(matches['boosters_usados'].describe())
```

file: purchases

```python
print(f'shape: {purchases.shape}')
print(purchases.dtypes)
print(purchases.isnull().sum())
display(purchases.head())

# Todo: Standardize categories
print(purchases['moneda'].value_counts())
print(purchases['plataforma'].value_counts())

# Describe
print(purchases['reembolso'].describe())
```

file: sessions

```python
print(f'shape: {sessions.shape}')
print(sessions.dtypes)
print(sessions.isnull().sum())
display(sessions.head())

# Todo: Standardize categories
print(sessions['plataforma'].value_counts())

# Describe
print(sessions['duracion_segundos'].describe())
atypical = sessions[sessions['duracion_segundos'].isin([-30,999999])]
print(f'atypical values: {atypical.shape}')
```

file: ua_spend

```python
print(f'shape: {ua_spend.shape}')
print(ua_spend.dtypes)
print(ua_spend.isnull().sum())
display(ua_spend.head())

# Todo: Standardize categories
print(ua_spend['canal'].value_counts())

# Describe 
print(ua_spend[['instalaciones','impresiones', 'clics']].describe())
```

file: users

```python
print(f'shape: {users.shape}')
print(users.dtypes)
print(users.isnull().sum())
display(users.head())
print(f'CLIENTES ÚNICOS: {users['user_id'].nunique()}')

# Todo: Standardize categories
print(users['canal_adquisicion'].value_counts())
print(users['pais'].value_counts())
print(users['plataforma'].value_counts())

# nulls values
null = users[users['version_app'].isnull()]
display(null)
```
